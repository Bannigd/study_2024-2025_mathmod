```julia
using DifferentialEquations, Plots;
```

```julia
function model1(u, p, t)
    x, y = u
    a, b, c, h = p
    dx = -a*x - b*y + abs(sin(t+2))
    dy = -c*x - h*y + abs(cos(t+1))
    return [dx, dy]
end
    
function model2(u, p, t)
    x, y = u
    a, b, c, h = p
    dx = -a*x - b*y + abs(sin(2*t))
    dy = -c*x*y - h*y + abs(cos(t))
    return [dx, dy]
end
```

```julia
u0 = [222000, 229000]
# a,b,c,h
params_1 = [0.223, 0.774, 0.665, 0.332]
params_2 = [0.291, 0.865, 0.456, 0.789]

t = (0, 1)
```

```julia
prob1 = ODEProblem(model1, u0, t, params_1)
```

```julia
sol_1 = solve(prob1)
```

```julia
plot(sol_1, label=["Армия X" "Армия Y"], xaxis="Время", yaxis="Численность")
savefig("/home/dastarikov/work/study/2024-2025/Математическое моделирование/mathmod/labs/lab3/report/image/jl_plot1.png")
```

```julia
prob2 = ODEProblem(model2, u0, t, params_2)
sol_2 = solve(prob2)
plot(sol_2, label=["Армия X" "Армия Y"], xaxis="Время", yaxis="Численность")
savefig("/home/dastarikov/work/study/2024-2025/Математическое моделирование/mathmod/labs/lab3/report/image/jl_plot2.png")
```

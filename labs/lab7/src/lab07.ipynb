```julia
using OrdinaryDiffEq, Plots
```

```julia
function model(u, p, t)
    N = 1430
    n = u
    a1, a2 = p
    dn = (a1 + a2*n)*(N-n)
    return dn
end
```

```julia
u0 = 2.
p1 = [0.62, 0.000032] # [a1, a2]
t = (0,10)
```

```julia
# уравнение 1
prob1 = ODEProblem(model, u0, t, p1)
sol1 = solve(prob1)
plot(sol1, label="n", xaxis="Время", yaxis="Количество")
savefig("/home/dastarikov/work/study/2024-2025/study_2024-2025_mathmod/labs/lab7/report/image/jl_1.png")
```

```julia
# уравнение 2
p2 = [0.000024, 0.4]
t = (0,.2)
prob2 = ODEProblem(model, u0, t, p2)
sol2 = solve(prob2)
plot(sol2, label="n", xaxis="Время", yaxis="Количество")
savefig("/home/dastarikov/work/study/2024-2025/study_2024-2025_mathmod/labs/lab7/report/image/jl_2.png")
```

```julia
using OrdinaryDiffEq, Plots
```

```julia
function model1(u, p, t)
    # SIR with I0>I*
    S, I, R = u
    alpha, beta = p
    dS = -alpha*S
    dI = alpha*S-beta*I
    dR = beta*I
    return [dS, dI, dR]
end
function model2(u, p, t)
    # SIR with I0<=I*
    S, I, R = u
    alpha, beta = p
    dS = 0
    dI = -beta*I
    dR = beta*I
    return [dS, dI, dR]

end
```

```julia
I0 = 100
R0 = 20
S0 = 9654 - I0 - R0
u0 = [S0, I0, R0]
p = [0.01, 0.02]
t = (0,700)
```

```julia
prob1 = ODEProblem(model1, u0, t, p)
prob2 = ODEProblem(model2, u0, t, p)
```

```julia
sol1 = solve(prob1)
sol2 = solve(prob2)
```

```julia
plot(sol1, label=["S" "I" "R"], xaxis="Время", yaxis="Численность")
#savefig("/home/dastarikov/work/study/2024-2025/study_2024-2025_mathmod/labs/lab6/report/image/jl_1.png")
```

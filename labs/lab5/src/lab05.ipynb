```julia
using OrdinaryDiffEq, Plots
```

```julia
function model(u, p, t)
    # Lottka-Volterra model
    x, y = u
    a, b, c, d = p
    dx = a*x + b*x*y
    dy = c*y + d*x*y
    return [dx, dy]
end
```

```julia
u0 = [4,12]
p = [-0.71, 0.046, 0.64, -0.017]
t = (0,100)
```

```julia
prob = ODEProblem(model, u0, t, p)
```

```julia
sol = solve(prob, reltol = 1e-12)
```

```julia
plot(sol, label=["X" "Y"], xaxis="Время", yaxis="Численность")
savefig("/home/dastarikov/work/study/2024-2025/study_2024-2025_mathmod/labs/lab5/report/image/jl_over_t.png")
```
